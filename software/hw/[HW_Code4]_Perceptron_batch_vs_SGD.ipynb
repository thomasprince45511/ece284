```python
import torch
from torch.autograd import Variable
import torch.nn as nn
import torch.nn.functional as F

import torch.optim as optim
import matplotlib.pyplot as plt

class Net(nn.Module):
    def __init__(self):
        super(Net, self).__init__()
        self.fc1 = nn.Linear(2,1,bias=False)
    def forward(self, x):
        x = self.fc1(x)
        return x

net = Net()
print(net)


net.fc1.weight = torch.nn.Parameter(torch.tensor([[1., -1.]], requires_grad=True))

print(list(net.parameters()))

#input = torch.randn(1,2)
#out = net(input)


criterion = nn.MSELoss()
#def criterion(out, label):
#    return ((label - out)**2).mean()


optimizer = optim.SGD(net.parameters(), lr=0.01, momentum=0.5)
#optimizer = optim.Adam(net.parameters(), lr=0.005)


data = torch.tensor([[1.,3.], [2.,6.], [3.,9.]], dtype=torch.float)
target = torch.tensor([[1.],[5.],[13.]], dtype=torch.float)

hist = []

############## Batch GD based update ##############       
      
for epoch in range(300): 
    for i, current_data in enumerate(data):  
        optimizer.zero_grad()
        outputs = net(current_data)
        current_target = target[i]
        loss = criterion(outputs, current_target)
        loss.backward()
        hist.append(loss.detach())
        optimizer.step()
    print("Epoch {} - loss: {}".format(epoch, loss))
####################################################

### Test the trained network ###
for i, current_data in enumerate(data):
    out = net(current_data)  
    print("when x = {}, y = {}".format(current_data, out))
    
plt.plot(hist, label = "training curve")
```

#### Differences in SGD vs Batch Gradient Descent
+ Stochastic Gradient descent is much noiser than Batch Gradient Descent
+ In our dataset we have 3 sets of inputs and 3 targets. Each time we update the weight, we are optimizing for just 1 of the 3 sets of inputs. The loss due to that input goes down, but the loss due to the other inputs may go up. 
+ With Batch Gradient Descent we look the the total loss of all 3 points and the optimise for that. So with each optimizer step, our total loss should slowly go down
